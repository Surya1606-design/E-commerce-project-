```python
# Check whether the delivered orders have missing dates
missing_customer_delivery = orders[
    (orders["order_status"] == 'Delivered') & 
    (orders["order_delivered_customer_date"].isna())
]

print("Delivered orders missing customer delivery date:", len(missing_customer_delivery))
```

```python
missing_carrier_delivery = orders[
    (orders["order_status"] == "delivered") &
    (orders["order_delivered_carrier_date"].isna())
]

print("Delivered orders missing carrier delivery date:",len(missing_carrier_delivery))
```

```python
print(missing_carrier_delivery[[
    "order_id",
    "order_status",
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]])
```

```python
orders["delivery_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_purchase_timestamp"]
).dt.total_seconds() / 86400
```

```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

orders[date_columns] = orders[date_columns].apply(
    pd.to_datetime, errors = "coerce"
)

print(orders[date_columns].dtypes)
```

```python
orders["delivery_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_purchase_timestamp"]
).dt.total_seconds() / 86400

print(orders["delivery_days"])
```

```python
# Check the no of orders in each status
print(orders["order_status"].value_counts(dropna=False))
```

```python
# Analyse missing dates by order status
missing_analysis = orders.groupby("order_status").agg(
    total_orders = ("order_id", "count"),
    missing_approved_at = ("order_approved_at",
                           lambda x: x.isna().sum()),
    missing_delivered_carrier = ("order_delivered_carrier_date",
                                 lambda x:x.isna().sum()),
    missing_delivered_customer = ("order_delivered_customer_date",
                                 lambda x: x.isna().sum())                     
)

print(missing_analysis)
```

```python
avearge_delivery_days = orders.loc[
    orders["order_status"].eq("delivered"),
    "delivery_days"
].mean()

print("Average delivery time", round(avearge_delivery_days, 2), "days")
```

```python
# Calculate delivery delay in days 
orders["delivery_delay_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_estimated_delivery_date"]
).dt.total_seconds() / 86400
print(orders["delivery_delay_days"])
```
